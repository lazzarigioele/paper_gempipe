```python
import shutil
import glob
import os

import pandas as pnd
from Bio import SeqIO, SeqRecord, SeqFeature, Seq
```

# 01_klebsiella

```python
# prepare the 37 input genomes:
# mamba update openpyxl
metadata = pnd.read_csv('01_klebsiella/tables/metadata.csv', index_col=0)

metadata['full_name'] = metadata['Species'] + ' '+ metadata['Strain']
metadata['full_name'] = metadata['full_name'].str.replace('K. ', 'Klebsiella ')
metadata.index = metadata.index + 1
metadata = metadata.rename(columns={i: i.lower() for i in metadata.columns})
metadata[['full_name', 'accession']].to_excel('01_klebsiella/tables/genomes.xlsx')

print(metadata.shape)
metadata
```

```python
# create a folder containing the input genomes, 
# having as filename the strain ID instead of the assemlby accession.

for index, row in metadata.iterrows():
    # get the assembly file
    if row['accession'] == 'NC_022566.1 + NC_005249.1' : 
        assembly = '01_klebsiella/genomes/NC_022566.1_NC_005249.1.fasta'
    else: assembly = '01_klebsiella/genomes/' + row['accession'] + '.fna'
    # get the strain name
    species = row['species']
    if   species == 'K. africana': species = 'Ka'
    elif species == 'K. pneumoniae': species = 'Kp'
    elif species == 'K. quasipneumoniae subsp quasipneumoniae': species = 'Kqq'
    elif species == 'K. quasipneumoniae subsp similipneumoniae': species = 'Kqs'
    elif species == 'K. quasivariicola': species = 'Kqv'
    elif species == 'K. variicola subsp tropica': species = 'Kvt'
    elif species == 'K. variicola subsp variicola': species = 'Kvv'
    strain = str(row['strain'])
    if ' (=' in strain: strain = strain.split(' (=', 1)[0]
    strain  = strain.replace(' ', '').replace('-', '_')
    shutil.copyfile(assembly, f'01_klebsiella/genomes_renamed/{species}_{strain}.fna')
```

# 02_ralstonia

```python
metadata = pnd.read_csv('02_ralstonia/tables/metadata.csv', index_col=0)
metadata.index = metadata.reset_index(drop=True).index + 1
metadata = metadata.rename(columns={i: i.lower() for i in metadata.columns})

metadata['full_name'] = metadata['species'] + ' ' + metadata['strain']
metadata['full_name'] = metadata['full_name'].str.replace('R. ', 'Ralstonia ')

metadata[['full_name', 'accession']].to_excel('02_ralstonia/tables/genomes.xlsx')
metadata
```

```python
# create a folder containing the input genomes, 
# having as filename the strain ID instead of the assemlby accession.

for file in glob.glob('02_ralstonia/genomes/*fna'):
    accession = file.replace('02_ralstonia/genomes/', '')
    accession = '_'.join(accession.split('_')[:2])
    match = metadata[metadata['accession']==accession].iloc[0]
    renamed = match['strain'] + '.fna'
    shutil.copyfile(file, '02_ralstonia/genomes_renamed/' + renamed)
```

# 04_streptococcus


```python
SD1 = pnd.read_excel('04_streptococcus/tables/41564_2022_1208_MOESM4_ESM.xlsx', sheet_name='Supplementary_data_1')
SD1.head()
```

```python
high_saturation = SD1[SD1['Tn-Seq libraries saturation (%)3']>35]
high_saturation = high_saturation.set_index('GenBank accession', drop=True)
high_saturation
```

```python
SD6 = pnd.read_excel('04_streptococcus/tables/41564_2022_1208_MOESM4_ESM.xlsx', sheet_name='Supplementary_data_6')
SD6.head()
```

```python


os.makedirs('04_streptococcus/proteomes_renamed/', exist_ok=True)


metadata = []
proteomes_raw = glob.glob('04_streptococcus/proteomes/*.1.faa')
for file in proteomes_raw:
    accession_ori = file.replace('04_streptococcus/proteomes/', '').replace('.1.faa', '')
    accession = accession_ori + '.1'
    strain_lab_name = high_saturation.loc[accession_ori, 'Common lab name']
    full_name = 'Streptococcus pneumoniae ' + strain_lab_name
    metadata.append({'full_name': full_name, 'accession': accession})
    output_file = f'04_streptococcus/proteomes_renamed/{strain_lab_name}.faa'
    records = []
    ids_added = set()
    for i, record in enumerate(SeqIO.parse(file, "fasta")):
        new_id = record.description.split('[locus_tag=',1)[-1].split(']', 1)[0]
        ids_added.add(new_id)
        record.id = new_id
        record.description = ""  
        
        # handle eventual asterisks and dashes:
        cnt_asterisks = record.seq.count("*")
        cnt_dashes = record.seq.count("-")
        record.seq = record.seq.rstrip("*")
        record.seq = record.seq.replace("*", "X")
        record.seq = record.seq.replace("-", '')
        records.append(record)
    
    SeqIO.write(records, output_file, "fasta")
    mentioned = set(SD6[SD6['Genome']==accession_ori]['locus_tag'].to_list())
    missing = mentioned - set(ids_added)
    print(f"Written '{output_file}'. Mentioned: {len(mentioned)}. Added: {len(ids_added)}. Missing sequences: {len(missing)}.")
    
metadata = pnd.DataFrame.from_records(metadata)
```

```python
os.makedirs('04_streptococcus/tables/', exist_ok=True)
metadata[['full_name', 'accession']].to_excel('04_streptococcus/tables/genomes.xlsx')
metadata
```

# 05 / 06 pseudomonas 


```python


suppl_table_03 = pnd.read_excel('05_pseudomonas/tables/pnas.1900570116.sd03.xlsx')
suppl_table_03 = suppl_table_03.set_index('Gene ID', drop=True, verify_integrity=True)
# format strain names
for index, row in suppl_table_03.iterrows():
    if row['Strain'] == 'BWH005': suppl_table_03.loc[index, 'Strain'] = 'BWHPSA005'
    if row['Strain'] == 'BWH013': suppl_table_03.loc[index, 'Strain'] = 'BWHPSA013'
    if row['Strain'] == 'BWH015': suppl_table_03.loc[index, 'Strain'] = 'BWHPSA015'
    if row['Strain'] == 'PA14': suppl_table_03.loc[index, 'Strain'] = 'UCBPP-PA14'

suppl_table_03
```

```python


accession_to_strain = {
    'GCA_000481765.1': '19660',
    'GCA_000480685.1': 'BL23',
    'GCA_000481605.1': 'BWHPSA005',
    'GCA_000481445.1': 'BWHPSA013',
    'GCA_000481405.1': 'BWHPSA015',
    'GCA_000480375.1': 'CF77',
    'GCA_000629365.1': 'PS75',
    'GCA_000014625.1': 'UCBPP-PA14',
    'GCA_000481705.1': 'X13273',
}

os.makedirs('05_pseudomonas/proteomes_renamed/', exist_ok=True)


metadata = []
for accession, strain in accession_to_strain.items():
    full_name = "Pseudomonas aeruginosa " + strain
    metadata.append({'full_name': full_name, 'accession': accession})
    
    ids_tested = set(list(suppl_table_03[suppl_table_03['Strain']==strain].index))
    
    infile = f"05_pseudomonas/proteomes/{accession}.gbff"
    sr_list = []
    ids_found = set()

    
    # use SeqIO.read() if you expect only one record in the GenBank file.
    # alternatively, use SeqIO.parse() if there are multiple records in the file.
    records = SeqIO.parse(infile, 'genbank')
    for record in records:

        # get scaffold information: 
        sequence_id = record.id
        sequence_description = record.description
        sequence_seqlen = len(record.seq)
        
        
        for feature in record.features:
            if feature.type == 'CDS':
                try: locus_tag = feature.qualifiers['locus_tag'][0]
                except: locus_tag = None
                try: aa_seq = feature.qualifiers['translation'][0]
                except: aa_seq = None
                
                if locus_tag not in ids_found:
                    sr = SeqRecord.SeqRecord(Seq.Seq(aa_seq), id=locus_tag, description='')
                    sr_list.append(sr)
                    ids_found.add(locus_tag)
                
    outfile = f'05_pseudomonas/proteomes_renamed/{strain}.faa'
    with open(outfile, 'w') as w_handler:
        count = SeqIO.write(sr_list, w_handler, "fasta")
        
        missing = ids_tested - ids_found
        print(f"Written '{outfile}'. Mentioned: {len(ids_tested)}. Added: {len(ids_found)}. Missing sequences: {len(missing)}.")
    
    
metadata = pnd.DataFrame.from_records(metadata)
```

```python
os.makedirs('05_pseudomonas/tables/', exist_ok=True)
metadata[['full_name', 'accession']].to_excel('05_pseudomonas/tables/genomes.xlsx')
os.makedirs('06_pseudomonas/tables/', exist_ok=True)
metadata[['full_name', 'accession']].to_excel('06_pseudomonas/tables/genomes.xlsx')
metadata
```
