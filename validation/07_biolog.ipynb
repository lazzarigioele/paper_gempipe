```python
# define the 94 substrates analyzed by Hawkey2022
# (names are according to biolog_mappings)

hawkey_substrates = set(list(biolog_exp['substrate'].unique()))
print('hawkey_substrates:', len(hawkey_substrates))
```

```python
# adjust strain names:
for index, row in biolog_exp.iterrows(): 
    strain_id = row['strain']
    strain_id = strain_id.replace('K_variicola_tropicalensis_', 'Kvt_')
    strain_id = strain_id.replace('K_pneumoniae_', 'Kp_')
    strain_id = strain_id.replace('K_quasi_simil_', 'Kqs_')
    strain_id = strain_id.replace('K_africa_', 'Ka_')
    strain_id = strain_id.replace('K_quasi_quasi_', 'Kqq_')
    strain_id = strain_id.replace('K_quasivariicola_', 'Kqv_')
    strain_id = strain_id.replace('K_variicola_variicola_', 'Kvv_')
    
    strain_id = strain_id.replace('CDC4241-71', 'CDC4241_71')
    strain_id = strain_id.replace('NJST258-1', 'NJST258_1')
    strain_id = strain_id.replace('NTUH_K2044', 'NTUHK2044')
    strain_id = strain_id.replace('CIP52_145', 'CIP52.145')
    strain_id = strain_id.replace('03-9138-2', '03_9138_2')
    strain_id = strain_id.replace('At-22', 'At_22')
    
    biolog_exp.loc[index, 'strain']  = strain_id

    
print(biolog_exp.shape)
biolog_exp.to_csv('01_klebsiella/tables/biolog_exp.csv')
print(len(biolog_exp['substrate'].unique()))
biolog_exp
```

## define main functions

```python
def get_biolog_results_df(files):
    
    
    results_df = [] 
    for file in files:
        
        # get the strain ID:
        strain_id = Path(file).stem
        strain_id = strain_id.replace('_model', '')
        if '-draft' in strain_id: 
            continue    # Skip gapseq draft models

            
        # load the simulation:
        df = pnd.read_csv(file, index_col=0)


        # simulate deletions
        for index, row in df.iterrows(): 
            
            if row['source'] == 'C':
                
                if   row['status'] == 'infeasible': sim = 'infeasible'
                elif row['exr_after_present'] and row['value'] > 0.001: sim = True
                else: sim = False
                    
                
                results_df.append({
                    'strain': strain_id,
                    'substrate': row['substrate'],
                    'source': row['source'],
                    'sim': sim,
                    'exr_before_present': row['exr_before_present'],
                    'exr_after_present': row['exr_after_present'],
                    'value': row['value'],
                    'status': row['status'],
                })

            
        print('▯', end='')  # loading bar

    results_df = pnd.DataFrame.from_records(results_df)
    return results_df
```

# define match

```python
biolog_sims = pnd.read_csv(f'tables/biolog_sims.csv', index_col=0)
biolog_sims
```

```python
biolog_exps = pnd.read_csv(f'tables/biolog_exps.csv', index_col=0)

# are  they all T/F ?
print(biolog_exps['growth_call'].unique())
# remove trials for which there's no clear experimental result:
biolog_exps = biolog_exps[biolog_exps['growth_call'].isna()==False]

# reset index for quicker search
biolog_exps['new_index'] = biolog_exps['strain'] + biolog_exps['substrate']
biolog_exps = biolog_exps.set_index('new_index', verify_integrity=True, drop=True)
biolog_exps
```

```python
# create a match dataframe:
match_df = biolog_sims.copy()


match_df['experimental'] = None
match_df['call'] = None

for index, row in match_df.iterrows(): 
    
    try: exp = biolog_exps.loc[row['strain']+row['substrate'], 'growth_call']
    except: exp = 'unknown'
    match_df.loc[index, 'experimental'] = exp
    
    sim = row['sim']
    if exp=='unknown': match_df.loc[index, 'call'] = 'unknown'
    else:
        if sim=='infeasible': sim = False
        else: sim = eval(sim)
        
        if   sim==True and exp==True:   match_df.loc[index, 'call'] = 'TP'
        elif sim==False and exp==False: match_df.loc[index, 'call'] = 'TN'
        elif sim==True and exp==False:  match_df.loc[index, 'call'] = 'FP'
        elif sim==False and exp==True:  match_df.loc[index, 'call'] = 'FN'

    match_df.loc[index, 'BiGG'] = biolog_mappings.loc[row['substrate'], 'BiGG_ID']
    match_df.loc[index, 'SEED'] = biolog_mappings.loc[row['substrate'], 'SEED_ID']
match_df
```

```python
# remove substrates having no BiGG nor SEED equivalent (eg 'Tween 20')
match_df = match_df[~(match_df['BiGG'].isna() & match_df['BiGG'].isna())]
match_df = match_df.reset_index(drop=True)
match_df
```
