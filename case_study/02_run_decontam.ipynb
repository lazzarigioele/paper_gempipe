```python
import os
import shutil
import glob
import pathlib


os.makedirs('genomes_decontam', exist_ok=True)
decontam = [pathlib.Path(g).stem for g in glob.glob('cocoremover/input/*.fna')]
for g in glob.glob('genomes_all/genbank/bacteria/*/*.fna'):
    acc = g.rsplit('/',2)[-2]
    if acc in decontam: 
        shutil.copyfile(f'cocoremover/output/{acc}.OK.fna', f'genomes_decontam/{acc}.fna')
    else:
        shutil.copyfile(g, f'genomes_decontam/{acc}.fna')
```
